```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv(r"C:\Users\91835\Downloads\archive\clean_final_data.csv")
df=df.head(50)
df
```

```python
plt.bar(df['Category'],df['Sales'],width=0.5) #Which product categories generate the highest total sales revenue?
plt.title('Category vs Sales')
plt.xlabel('Category of the Products')
plt.ylabel('No of Sales')
```

```python
#How does discount percentage affect sales value?
plt.figure(figsize=(10,6))
plt.scatter(df['Discount'],df['Sales'],color='red',marker='.')
plt.xlabel('Discount %')
plt.ylabel('No of sales')
plt.title('How does discount percentage affect sales value?')
```

```python
#What is the distribution of customer ages?
plt.hist(df['Age'])
plt.xlabel('Age Groups')
plt.ylabel('No of peoples')
```

```python
#Which customer segment (New, Regular, VIP) contributes the most sales?
plt.bar(df['CustomerSegment'],df['Sales'],width=0.5)
```

```python
#How does customer age influence order value?
plt.hist(df['OrderValue'])
```

```python
#Which customer segment uses each payment method most frequently?

P_segment=pd.crosstab(df['CustomerSegment'],df['PaymentMethod'])
plt.bar(np.arange(P_segment.shape[0])-0.3,P_segment['Cash'],label='Cash',width=0.2)
plt.bar(np.arange(P_segment.shape[0])-0.2,P_segment['Wallet'],label='Wallet',width=0.2)
plt.bar(np.arange(P_segment.shape[0]),P_segment['CardToCard'],label='CardToCard',width=0.2)
plt.bar(np.arange(P_segment.shape[0])+0.2,P_segment['Gateway'],label='Gateway',width=0.2)
plt.xticks(np.arange(P_segment.shape[0]),P_segment.index)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

payment_counts = df['PaymentMethod'].value_counts()

plt.pie(
    payment_counts,
    labels=payment_counts.index,
    autopct='%1.1f%%'
)

plt.title('Payment Method Distribution')
plt.show()
```
